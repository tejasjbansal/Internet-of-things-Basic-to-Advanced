## Step 2 — Import Libraries

```

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

from sklearn.ensemble import IsolationForest
from sklearn.preprocessing import MinMaxScaler

from datetime import datetime, timedelta
import random
```

## Step 3 — Generate Synthetic IoT Sensor Data

```

data = []

start_time = datetime.now()

for i in range(300):

    timestamp = start_time + timedelta(seconds=i)

    temperature = round(np.random.normal(30, 3), 2)
    humidity = round(np.random.normal(60, 5), 2)
    pressure = round(np.random.normal(1013, 5), 2)
    vibration = round(np.random.normal(4, 0.8), 2)
    current = round(np.random.normal(10, 1), 2)

    # Inject anomalies
    if random.random() < 0.08:
        temperature += random.randint(10, 20)
        vibration += random.randint(3, 6)
        current += random.randint(4, 8)

    data.append([
        timestamp,
        temperature,
        humidity,
        pressure,
        vibration,
        current
    ])

df = pd.DataFrame(data, columns=[
    'timestamp',
    'temperature',
    'humidity',
    'pressure',
    'vibration',
    'current'
])

df.head()
```

## Step 4 — Visualize Raw Sensor Data

```

plt.figure(figsize=(12,5))

plt.plot(df['temperature'])

plt.title("Raw Temperature Sensor Data")
plt.xlabel("Reading")
plt.ylabel("Temperature")

plt.show()
```

## Step 5 — Normalize Sensor Data

```

scaler = MinMaxScaler()

sensor_columns = [
    'temperature',
    'humidity',
    'pressure',
    'vibration',
    'current'
]

df_scaled = df.copy()

df_scaled[sensor_columns] = scaler.fit_transform(df[sensor_columns])

df_scaled.head()
```

## Step 6 — Prepare Features for Anomaly Detection

```

X = df_scaled[sensor_columns]

X.head()
```

## Step 7 — Train Isolation Forest Model

```

model = IsolationForest(
    contamination=0.08,
    random_state=42
)

model.fit(X)

print("Anomaly Detection Model Trained")
```

## Step 8 — Detect Anomalies

```

df['anomaly'] = model.predict(X)

df['anomaly'] = df['anomaly'].map({
    1: 0,
    -1: 1
})

df[['temperature', 'vibration', 'current', 'anomaly']].head(20)
```

## Step 9 — Count Detected Anomalies

```

anomaly_count = df['anomaly'].sum()

print("Total anomalies detected:", anomaly_count)
```

## Step 10 — Visualize Detected Anomalies

```

plt.figure(figsize=(14,6))

plt.plot(df['temperature'], label='Temperature')

anomalies = df[df['anomaly'] == 1]

plt.scatter(
    anomalies.index,
    anomalies['temperature'],
    label='Anomaly'
)

plt.title("Anomaly Detection in Temperature Sensor Data")
plt.xlabel("Reading")
plt.ylabel("Temperature")
plt.legend()

plt.show()
```

## Step 11 — Simulate Real-Time AIoT Monitoring

```

new_sensor_data = pd.DataFrame({
    'temperature': [0.92],
    'humidity': [0.55],
    'pressure': [0.60],
    'vibration': [0.95],
    'current': [0.90]
})

prediction = model.predict(new_sensor_data)

if prediction[0] == -1:
    print("ALERT: Anomaly Detected in AIoT System")
else:
    print("System Operating Normally")
```
